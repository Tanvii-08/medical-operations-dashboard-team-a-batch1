```python
import pandas as pd

lab = pd.read_csv("lab_results_clean.csv")
billing = pd.read_csv("billing_clean.csv")

print(lab.head())
print(billing.head())
```

```python
# Treatment / Service Demand

test_demand = lab["Test"].value_counts()

print(test_demand)
```

```python
# Merge lab results with billing data using Admission_ID

merged = pd.merge(
    lab,
    billing,
    on="Admission_ID",
    how="inner"
)

print(merged.head())
print("Number of records:", len(merged))
```

```python
# Billing by laboratory test

billing_by_test = (
    merged.groupby("Test")["Total"]
    .agg(["count", "sum", "mean"])
    .sort_values("sum", ascending=False)
)

print(billing_by_test)
```

```python
# Overall Billing KPIs

total_billing = billing["Total"].sum()
total_insurance = billing["Insurance"].sum()
total_paid = billing["Paid"].sum()
total_pending = billing["Pending"].sum()

print("Total Billing:", f"{total_billing:,}")
print("Total Insurance:", f"{total_insurance:,}")
print("Total Paid:", f"{total_paid:,}")
print("Total Pending:", f"{total_pending:,}")
```

```python
# Payment and Pending Rates

paid_percentage = (total_paid / total_billing) * 100
pending_percentage = (total_pending / total_billing) * 100

print("Paid Percentage:", round(paid_percentage, 2), "%")
print("Pending Percentage:", round(pending_percentage, 2), "%")
```

```python
# Member 7 - Treatment & Service Demand KPI Summary

# Most demanded test
most_demanded_test = lab["Test"].value_counts().idxmax()
most_demanded_count = lab["Test"].value_counts().max()

# Test demand
test_demand = lab["Test"].value_counts()

# Billing by test
billing_by_test = (
    merged.groupby("Test")["Total"]
    .agg(["count", "sum", "mean"])
    .sort_values("sum", ascending=False)
)

# Overall billing
total_billing = billing["Total"].sum()
total_paid = billing["Paid"].sum()
total_pending = billing["Pending"].sum()

# Payment percentages
paid_percentage = (total_paid / total_billing) * 100
pending_percentage = (total_pending / total_billing) * 100


# Display results
print("===== TREATMENT & SERVICE DEMAND =====")
print()

print("Most Demanded Test:", most_demanded_test)
print("Number of Tests:", f"{most_demanded_count:,}")
print()

print("Blood Sugar Tests:", f"{test_demand['Blood Sugar']:,}")
print("Cholesterol Tests:", f"{test_demand['Cholesterol']:,}")
print()

print("Blood Sugar Billing: ₹", f"{billing_by_test.loc['Blood Sugar', 'sum']:,}")
print("Cholesterol Billing: ₹", f"{billing_by_test.loc['Cholesterol', 'sum']:,}")
print()

print("Total Billing: ₹", f"{total_billing:,}")
print("Total Paid: ₹", f"{total_paid:,}")
print("Total Pending: ₹", f"{total_pending:,}")
print()

print("Paid Percentage:", round(paid_percentage, 2), "%")
print("Pending Percentage:", round(pending_percentage, 2), "%")


```
