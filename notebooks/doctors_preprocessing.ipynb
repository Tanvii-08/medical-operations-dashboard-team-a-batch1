```python
# ==========================================================
# Remove leading and trailing spaces from string columns
# (Keeps the dataset standardized for current and future data)
# ==========================================================

string_columns = ["Doctor_ID", "Doctor_Name", "Department_ID", "Specialization"]

for col in string_columns:
    doctors[col] = doctors[col].str.strip()
```
