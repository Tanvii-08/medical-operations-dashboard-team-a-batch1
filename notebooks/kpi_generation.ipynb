```python
import pandas as pd

patients = pd.read_csv("data/raw/patients.csv")
admissions = pd.read_csv("data/raw/admissions.csv")
billing = pd.read_csv("data/raw/billing.csv")
doctors = pd.read_csv("data/raw/doctors.csv")
departments = pd.read_csv("data/raw/departments.csv")
lab_results = pd.read_csv("data/raw/lab_results.csv")
medications = pd.read_csv("data/raw/medications.csv")
surgeries = pd.read_csv("data/raw/raw_surgeries.csv")
```

```python
# ===============================
# Healthcare Operational KPIs
# ===============================

print("========== HOSPITAL OPERATIONAL KPIs ==========\n")

# Patient KPIs
print("Total Patients:", patients["Patient_ID"].nunique())
print("Average Patient Age:", round(patients["Age"].mean(), 2))
print("Male Patients:", (patients["Gender"] == "Male").sum())
print("Female Patients:", (patients["Gender"] == "Female").sum())

print("\n---------------------------------\n")

# Admission KPIs
print("Total Admissions:", admissions["Admission_ID"].nunique())
print("Active Admissions:", (admissions["Status"] == "Admitted").sum())
print("Discharged Patients:", (admissions["Status"] == "Discharged").sum())

print("\n---------------------------------\n")

# Doctor KPIs
print("Total Doctors:", doctors["Doctor_ID"].nunique())
print("Average Doctor Experience:", round(doctors["Experience"].mean(), 2), "Years")

print("\n---------------------------------\n")

# Department KPIs
print("Total Departments:", departments["Department_ID"].nunique())

print("\n---------------------------------\n")

# Billing KPIs
print("Total Revenue:", billing["Total"].sum())
print("Insurance Amount:", billing["Insurance"].sum())
print("Paid Amount:", billing["Paid"].sum())
print("Pending Amount:", billing["Pending"].sum())

collection_rate = (billing["Paid"].sum() / billing["Total"].sum()) * 100
print("Collection Rate:", round(collection_rate, 2), "%")

print("\n---------------------------------\n")

# Surgery KPIs
print("Total Surgeries:", surgeries["Surgery_ID"].nunique())

success_rate = surgeries["Success_Flag"].mean() * 100
print("Surgery Success Rate:", round(success_rate, 2), "%")

print("Average Surgery Duration:", round(surgeries["Duration_Minutes"].mean(), 2), "minutes")
print("Average Surgery Cost:", round(surgeries["Cost"].mean(), 2))
```

```python
dept_admissions = admissions.merge(departments, on="Department_ID")

dept_kpi = dept_admissions.groupby("Department_Name").size().sort_values(ascending=False)

print(dept_kpi)
```

```python
billing_dept = admissions.merge(billing, on="Admission_ID")
billing_dept = billing_dept.merge(departments, on="Department_ID")

revenue = billing_dept.groupby("Department_Name")["Total"].sum().sort_values(ascending=False)

print(revenue)
```

```python
doctor_count = doctors.merge(departments, on="Department_ID")

doctor_count = doctor_count.groupby("Department_Name")["Doctor_ID"].count().sort_values(ascending=False)

print(doctor_count)
```

```python
kpi_summary = pd.DataFrame({

"Category":[
"Patient","Patient","Patient","Patient",
"Admission","Admission","Admission","Admission","Admission",
"Doctor","Doctor","Doctor",
"Department","Department",
"Finance","Finance","Finance","Finance","Finance","Finance",
"Surgery","Surgery","Surgery","Surgery",
"Laboratory",
"Medication"
],

"KPI":[
"Total Patients",
"Average Patient Age",
"Male Patients",
"Female Patients",

"Total Admissions",
"Patients Under Treatment",
"Recovered Patients",
"Discharged Patients",
"Average Admissions per Patient",

"Total Doctors",
"Average Doctor Experience (Years)",
"Doctors per Department",

"Total Departments",
"Average Admissions per Department",

"Total Revenue",
"Insurance Amount",
"Paid Amount",
"Pending Amount",
"Collection Rate (%)",
"Insurance Coverage (%)",

"Total Surgeries",
"Surgery Success Rate (%)",
"Average Surgery Duration (Minutes)",
"Average Surgery Cost",

"Total Lab Records",

"Total Medication Records"
],

"Current Value":[

# Patient
patients["Patient_ID"].nunique(),
round(patients["Age"].mean(),2),
(patients["Gender"]=="Male").sum(),
(patients["Gender"]=="Female").sum(),

# Admission
admissions["Admission_ID"].nunique(),
admissions["Status"].isin(["Critical","Under Treatment"]).sum(),
(admissions["Status"]=="Recovered").sum(),
(admissions["Status"]=="Discharged").sum(),
round(admissions["Admission_ID"].nunique()/patients["Patient_ID"].nunique(),2),

# Doctor
doctors["Doctor_ID"].nunique(),
round(doctors["Experience"].mean(),2),
round(doctors["Doctor_ID"].nunique()/departments["Department_ID"].nunique(),2),

# Department
departments["Department_ID"].nunique(),
round(admissions["Admission_ID"].nunique()/departments["Department_ID"].nunique(),2),

# Finance
billing["Total"].sum(),
billing["Insurance"].sum(),
billing["Paid"].sum(),
billing["Pending"].sum(),
round((billing["Paid"].sum()/billing["Total"].sum())*100,2),
round((billing["Insurance"].sum()/billing["Total"].sum())*100,2),

# Surgery
surgeries["Surgery_ID"].nunique(),
round(surgeries["Success_Flag"].mean()*100,2),
round(surgeries["Duration_Minutes"].mean(),2),
round(surgeries["Cost"].mean(),2),

# Laboratory
len(lab_results),

# Medication
len(medications)

]

})

# Format values
def format_value(x):
    if isinstance(x,(int,float)):
        if float(x).is_integer():
            return f"{int(x):,}"
        else:
            return f"{x:,.2f}"
    return x

kpi_summary["Current Value"] = kpi_summary["Current Value"].apply(format_value)

kpi_summary
```
