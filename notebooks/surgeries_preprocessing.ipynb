```python
import pandas as pd
surgeries = pd.read_csv("surgeries_final.csv")
print("Initial shape:", surgeries.shape)
surgeries.head()
```

```python
surgeries.info()
```

```python
print("Full duplicate rows:", surgeries.duplicated().sum())
print("Duplicate Surgery_IDs:", surgeries["Surgery_ID"].duplicated().sum())
```

```python
print("Missing values per column:")
surgeries.isnull().sum()
```

```python
surgeries["Surgery_Date"] = pd.to_datetime(surgeries["Surgery_Date"])
surgeries.dtypes
```

```python
text_columns = ["Surgery_ID", "Admission_ID", "Patient_ID", "Doctor_ID", "Surgery_Type", "Outcome",
                "Surgeon_Name", "Assistant_Surgeon_Name", "Nurse_In_Charge", "Operation_Theatre_ID"]
for col in text_columns:
    surgeries[col] = surgeries[col].str.strip()
surgeries["Outcome"] = surgeries["Outcome"].str.title()
surgeries["Surgery_Type"] = surgeries["Surgery_Type"].str.title()
print("Unique Outcome values:", surgeries["Outcome"].unique())
```

```python
invalid_rows = surgeries[
    (surgeries["Duration_Minutes"] <= 0) |
    (surgeries["Cost"] <= 0) |
    (surgeries["Recovery_Days"] <= 0) |
    (surgeries["Length_Of_Stay_After_Surgery"] <= 0)
]
print("Rows with impossible Duration/Cost/Recovery/Stay values:", len(invalid_rows))
```

```python
surgeries["Total_Recovery_Days"] = (
    surgeries["Length_Of_Stay_After_Surgery"] + surgeries["Recovery_Days"]
)
surgeries["Total_Recovery_Days"].describe()
```

```python
surgeries.to_csv("./surgeries_clean.csv", index=False)
print("Saved cleaned file to data/processed/surgeries_clean.csv")
print("Final shape:", surgeries.shape)
```
