```python
import pandas as pd

# Load staff capacity dataset
staff_df = pd.read_csv("../data/raw/staff_capacity.csv")

staff_df.head()
```

```python
#

total_staff = staff_df["Total_Staff"].sum()
average_staff = staff_df["Total_Staff"].mean()
maximum_staff = staff_df["Total_Staff"].max()
minimum_staff = staff_df["Total_Staff"].min()

print("Total Staff:", total_staff)
print("Average Staff per Department:", round(average_staff, 2))
print("Maximum Staff in a Department:", maximum_staff)
print("Minimum Staff in a Department:", minimum_staff)
```

```python
# Highest and Lowest Staff Departments

highest_staff = staff_df.loc[staff_df["Total_Staff"].idxmax()]
lowest_staff = staff_df.loc[staff_df["Total_Staff"].idxmin()]

print("Department with Highest Staff:")
print(highest_staff["department_name"], "-", highest_staff["Total_Staff"])

print("\nDepartment with Lowest Staff:")
print(lowest_staff["department_name"], "-", lowest_staff["Total_Staff"])
```

```python
# Department-wise Staff Ranking

staff_ranking = staff_df.sort_values(
    by="Total_Staff",
    ascending=False
)

print(staff_ranking[["department_name", "Total_Staff"]])
```

```python
# Workforce Efficiency

staff_df["Staff_Share_Percent"] = (
    staff_df["Total_Staff"] / total_staff
) * 100

staff_efficiency = staff_df[
    ["department_name", "Total_Staff", "Staff_Share_Percent"]
].sort_values(
    by="Staff_Share_Percent",
    ascending=False
)

staff_efficiency["Staff_Share_Percent"] = (
    staff_efficiency["Staff_Share_Percent"].round(2)
)

print(staff_efficiency)
```

```python
# Patients per Staff

admissions_df = pd.read_csv("../data/raw/admissions.csv")

total_admissions = len(admissions_df)

patients_per_staff = total_admissions / total_staff

print("Total Admissions:", total_admissions)
print("Total Staff:", total_staff)
print("Patients per Staff:", round(patients_per_staff, 2))
```

```python
# Department-wise Patients per Staff

department_staff = admissions_df.groupby(
    "Department_ID"
).size().reset_index(name="Admissions")

department_staff = department_staff.merge(
    staff_df[["department_id", "department_name", "Total_Staff"]],
    left_on="Department_ID",
    right_on="department_id",
    how="left"
)

department_staff["Patients_per_Staff"] = (
    department_staff["Admissions"] /
    department_staff["Total_Staff"]
)

department_staff["Patients_per_Staff"] = (
    department_staff["Patients_per_Staff"].round(2)
)

department_staff = department_staff.sort_values(
    by="Patients_per_Staff",
    ascending=False
)

print(
    department_staff[
        [
            "Department_ID",
            "department_name",
            "Admissions",
            "Total_Staff",
            "Patients_per_Staff"
        ]
    ]
)
```

```python
# Final Workforce Efficiency Table

workforce_analysis = department_staff[
    [
        "Department_ID",
        "department_name",
        "Admissions",
        "Total_Staff",
        "Patients_per_Staff"
    ]
].copy()

workforce_analysis["Staff_Share_Percent"] = (
    workforce_analysis["Total_Staff"] / total_staff * 100
).round(2)

workforce_analysis = workforce_analysis.sort_values(
    by="Patients_per_Staff",
    ascending=False
)

workforce_analysis
```

```python
# Identify departments with high staffing pressure

high_pressure = workforce_analysis[
    workforce_analysis["Patients_per_Staff"] >
    workforce_analysis["Patients_per_Staff"].median()
]

high_pressure = high_pressure.sort_values(
    by="Patients_per_Staff",
    ascending=False
)

print("Departments above the median Patients-per-Staff ratio:")
print(
    high_pressure[
        [
            "Department_ID",
            "department_name",
            "Admissions",
            "Total_Staff",
            "Patients_per_Staff"
        ]
    ]
)
```

```python
# Final Workforce & Staffing Efficiency Output

final_workforce = workforce_analysis[
    [
        "Department_ID",
        "department_name",
        "Admissions",
        "Total_Staff",
        "Staff_Share_Percent",
        "Patients_per_Staff"
    ]
].copy()

final_workforce = final_workforce.sort_values(
    by="Patients_per_Staff",
    ascending=False
)

final_workforce
```

```python
# Save final workforce analysis for visualization

final_workforce.to_csv(
    "../data/processed/workforce_staffing_efficiency.csv",
    index=False
)

print("Workforce staffing analysis saved successfully.")
```
