# Workforce & Staffing Efficiency — Visualization

### Objective
Convert the Workforce & Staffing Efficiency analysis into interactive,
business-focused Plotly visualizations.

### Source
Member 3 — Workforce & Staffing Efficiency Analysis

### Key Metrics
- Total Staff
- Total Admissions
- Patients per Staff
- Staff Share by Department

### Business Focus
Identify departments with higher patients-per-staff workload and
understand how staff resources are distributed across departments.

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.express as px

from IPython.display import display, HTML
```

```python
df = pd.read_csv("../data/processed/workforce_staffing_efficiency.csv")

df.head()
```

```python
df.columns
```

```python
df["Admissions"] = pd.to_numeric(df["Admissions"])
df["Total_Staff"] = pd.to_numeric(df["Total_Staff"])
df["Staff_Share_Percent"] = pd.to_numeric(df["Staff_Share_Percent"])
df["Patients_per_Staff"] = pd.to_numeric(df["Patients_per_Staff"])

df = df.sort_values("Patients_per_Staff", ascending=False).reset_index(drop=True)

df.head()
```

```python
TOTAL_STAFF = df["Total_Staff"].sum()
TOTAL_ADMISSIONS = df["Admissions"].sum()

OVERALL_PATIENTS_PER_STAFF = (
    TOTAL_ADMISSIONS / TOTAL_STAFF
)

highest_pressure_row = df.loc[df["Patients_per_Staff"].idxmax()]

HIGHEST_PRESSURE_DEPT = highest_pressure_row["department_name"]
HIGHEST_PRESSURE_VALUE = highest_pressure_row["Patients_per_Staff"]

print("Total Staff:", TOTAL_STAFF)
print("Total Admissions:", TOTAL_ADMISSIONS)
print("Overall Patients per Staff:", round(OVERALL_PATIENTS_PER_STAFF, 2))
print(
    "Highest Staffing Pressure:",
    HIGHEST_PRESSURE_DEPT,
    "-",
    HIGHEST_PRESSURE_VALUE
)
```

```python
departments = ["All Departments"] + sorted(df["department_name"].unique().tolist())

department_dropdown = [
    {
        "label": "All Departments",
        "method": "update",
        "args": [
            {
                "visible": [True] * len(df)
            }
        ]
    }
]

print(departments)
```

```python
chart1_df = df.sort_values("Patients_per_Staff", ascending=True)

fig1 = px.bar(
    chart1_df,
    x="Patients_per_Staff",
    y="department_name",
    orientation="h",
    text="Patients_per_Staff",
    title="Patients per Staff by Department",
    labels={
        "Patients_per_Staff": "Patients per Staff",
        "department_name": "Department"
    }
)

fig1.update_traces(
    marker_color="#E67E22",
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig1.update_layout(
    height=550,
    margin=dict(l=20, r=40, t=70, b=40),
    title_x=0.5,
    plot_bgcolor="white",
    paper_bgcolor="white"
)

fig1.show()
```

```python
chart2_df = df.sort_values("Total_Staff", ascending=False)

fig2 = px.bar(
    chart2_df,
    x="department_name",
    y="Total_Staff",
    text="Total_Staff",
    title="Staff Distribution Across Departments",
    labels={
        "department_name": "Department",
        "Total_Staff": "Total Staff"
    }
)

fig2.update_traces(
    marker_color="#2F5597",
    textposition="outside"
)
fig2.update_layout(
    height=550,
    margin=dict(l=40, r=30, t=70, b=120),
    title_x=0.5,
    plot_bgcolor="white",
    paper_bgcolor="white",
    xaxis_tickangle=-45
)

fig2.show()
```

```python
fig3 = px.pie(
    df,
    names="department_name",
    values="Total_Staff",
    hole=0.55,
    title="Staff Share by Department"
)

fig3.update_traces(
    textposition="inside",
    textinfo="percent",
    marker=dict(
        line=dict(color="white", width=1)
    )
)

fig3.update_layout(
    height=550,
    title_x=0.5,
    margin=dict(l=20, r=20, t=70, b=30),
    paper_bgcolor="white"
)

fig3.show()
```

```python
fig4 = px.scatter(
    df,
    x="Total_Staff",
    y="Admissions",
    size="Patients_per_Staff",
    hover_name="department_name",
    hover_data={
        "Total_Staff": True,
        "Admissions": True,
        "Patients_per_Staff": ":.2f",
        "Staff_Share_Percent": ":.2f"
    },
    title="Admissions vs Total Staff",
    labels={
        "Total_Staff": "Total Staff",
        "Admissions": "Admissions"
    }
)

fig4.update_traces(
    marker=dict(
        color="#16A085",
        line=dict(width=1)
    )
)

fig4.show()
```

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px
import pandas as pd

# --------------------------------------------------
# 1. Load data
# --------------------------------------------------

dashboard_df = pd.read_csv(
    "../data/processed/workforce_staffing_efficiency.csv"
)

dashboard_df["Admissions"] = pd.to_numeric(
    dashboard_df["Admissions"]
)

dashboard_df["Total_Staff"] = pd.to_numeric(
    dashboard_df["Total_Staff"]
)

dashboard_df["Staff_Share_Percent"] = pd.to_numeric(
    dashboard_df["Staff_Share_Percent"]
)

dashboard_df["Patients_per_Staff"] = pd.to_numeric(
    dashboard_df["Patients_per_Staff"]
)


# --------------------------------------------------
# 2. Create Dash application
# --------------------------------------------------

app = Dash(__name__)

app.title = "Workforce & Staffing Efficiency"


# --------------------------------------------------
# 3. Dashboard Layout
# --------------------------------------------------

app.layout = html.Div(
    style={
        "fontFamily": "Arial, sans-serif",
        "backgroundColor": "#F4F6F8",
        "minHeight": "100vh",
        "padding": "20px"
    },

    children=[

        # ---------------- HEADER ----------------

        html.Div(
            style={
                "backgroundColor": "white",
                "padding": "18px 25px",
                "borderRadius": "12px",
                "marginBottom": "18px",
                "boxShadow": "0 2px 8px rgba(0,0,0,0.08)"
            },

            children=[
                html.H1(
                    "WORKFORCE & STAFFING EFFICIENCY",
                    style={
                        "margin": "0",
                        "fontSize": "26px",
                        "color": "#1F2937"
                    }
                ),

                html.P(
                    "Department-level staffing workload and resource distribution",
                    style={
                        "margin": "6px 0 0 0",
                        "color": "#6B7280",
                        "fontSize": "14px"
                    }
                )
            ]
        ),


        # ---------------- MAIN AREA ----------------

        html.Div(

            style={
                "display": "flex",
                "gap": "18px",
                "alignItems": "flex-start"
            },

            children=[

                # ==============================
                # LEFT FILTER PANEL
                # ==============================

                html.Div(

                    style={
                        "width": "220px",
                        "backgroundColor": "white",
                        "padding": "20px",
                        "borderRadius": "12px",
                        "boxShadow": "0 2px 8px rgba(0,0,0,0.08)",
                        "flexShrink": "0"
                    },

                    children=[

                        html.H3(
                            "FILTERS",
                            style={
                                "marginTop": "0",
                                "color": "#1F2937",
                                "fontSize": "17px"
                            }
                        ),

                        html.Label(
                            "Department",
                            style={
                                "fontWeight": "bold",
                                "fontSize": "13px",
                                "color": "#4B5563"
                            }
                        ),

                        dcc.Dropdown(
                            id="department-filter",

                            options=[
                                {
                                    "label": "All Departments",
                                    "value": "All Departments"
                                }
                            ]
                            +
                            [
                                {
                                    "label": dept,
                                    "value": dept
                                }
                                for dept in sorted(
                                    dashboard_df[
                                        "department_name"
                                    ].unique()
                                )
                            ],

                            value="All Departments",

                            clearable=False,

                            style={
                                "marginTop": "8px",
                                "fontSize": "13px"
                            }
                        ),

                        html.Hr(
                            style={
                                "margin": "25px 0"
                            }
                        ),

                        html.P(
                            "Use the department filter to explore staffing workload and resource distribution.",
                            style={
                                "fontSize": "12px",
                                "lineHeight": "1.5",
                                "color": "#6B7280"
                            }
                        )
                    ]
                ),


                # ==============================
                # RIGHT DASHBOARD AREA
                # ==============================

                html.Div(

                    style={
                        "flex": "1"
                    },

                    children=[

                        # ======================
                        # KPI CARDS
                        # ======================

                        html.Div(

                            style={
                                "display": "grid",
                                "gridTemplateColumns":
                                    "repeat(4, 1fr)",
                                "gap": "14px",
                                "marginBottom": "18px"
                            },

                            children=[

                                # KPI 1
                                html.Div(
                                    [
                                        html.P(
                                            "TOTAL STAFF",
                                            style={
                                                "fontSize": "12px",
                                                "color": "#6B7280",
                                                "margin": "0"
                                            }
                                        ),

                                        html.H2(
                                            id="kpi-staff",
                                            style={
                                                "color": "#2F5597",
                                                "margin": "8px 0 0 0"
                                            }
                                        )
                                    ],

                                    style={
                                        "backgroundColor": "white",
                                        "padding": "18px",
                                        "borderRadius": "10px",
                                        "borderLeft":
                                            "5px solid #2F5597",
                                        "boxShadow":
                                            "0 2px 8px rgba(0,0,0,0.08)"
                                    }
                                ),


                                # KPI 2
                                html.Div(
                                    [
                                        html.P(
                                            "TOTAL ADMISSIONS",
                                            style={
                                                "fontSize": "12px",
                                                "color": "#6B7280",
                                                "margin": "0"
                                            }
                                        ),

                                        html.H2(
                                            id="kpi-admissions",
                                            style={
                                                "color": "#E67E22",
                                                "margin": "8px 0 0 0"
                                            }
                                        )
                                    ],

                                    style={
                                        "backgroundColor": "white",
                                        "padding": "18px",
                                        "borderRadius": "10px",
                                        "borderLeft":
                                            "5px solid #E67E22",
                                        "boxShadow":
                                            "0 2px 8px rgba(0,0,0,0.08)"
                                    }
                                ),


                                # KPI 3
                                html.Div(
                                    [
                                        html.P(
                                            "PATIENTS / STAFF",
                                            style={
                                                "fontSize": "12px",
                                                "color": "#6B7280",
                                                "margin": "0"
                                            }
                                        ),

                                        html.H2(
                                            id="kpi-ratio",
                                            style={
                                                "color": "#16A085",
                                                "margin": "8px 0 0 0"
                                            }
                                        )
                                    ],

                                    style={
                                        "backgroundColor": "white",
                                        "padding": "18px",
                                        "borderRadius": "10px",
                                        "borderLeft":
                                            "5px solid #16A085",
                                        "boxShadow":
                                            "0 2px 8px rgba(0,0,0,0.08)"
                                    }
                                ),


                                # KPI 4
                                html.Div(
                                    [
                                        html.P(
                                            "HIGHEST PRESSURE",
                                            style={
                                                "fontSize": "12px",
                                                "color": "#6B7280",
                                                "margin": "0"
                                            }
                                        ),

                                        html.H2(
                                            id="kpi-pressure",
                                            style={
                                                "color": "#8E44AD",
                                                "margin": "8px 0 0 0",
                                                "fontSize": "20px"
                                            }
                                        ),

                                        html.P(
                                            id="kpi-pressure-value",
                                            style={
                                                "margin": "3px 0 0 0",
                                                "fontSize": "12px",
                                                "color": "#6B7280"
                                            }
                                        )
                                    ],

                                    style={
                                        "backgroundColor": "white",
                                        "padding": "18px",
                                        "borderRadius": "10px",
                                        "borderLeft":
                                            "5px solid #8E44AD",
                                        "boxShadow":
                                            "0 2px 8px rgba(0,0,0,0.08)"
                                    }
                                )
                            ]
                        ),


                        # ======================
                        # ROW 1 — TWO CHARTS
                        # ======================

                        html.Div(

                            style={
                                "display": "grid",
                                "gridTemplateColumns":
                                    "1fr 1fr",
                                "gap": "18px",
                                "marginBottom": "18px"
                            },

                            children=[

                                dcc.Graph(
                                    id="patients-staff-chart",
                                    style={
                                        "backgroundColor": "white"
                                    }
                                ),

                                dcc.Graph(
                                    id="staff-distribution-chart",
                                    style={
                                        "backgroundColor": "white"
                                    }
                                )
                            ]
                        ),


                        # ======================
                        # ROW 2 — TWO CHARTS
                        # ======================

                        html.Div(

                            style={
                                "display": "grid",
                                "gridTemplateColumns":
                                    "1fr 1fr",
                                "gap": "18px"
                            },

                            children=[

                                dcc.Graph(
                                    id="staff-share-chart"
                                ),

                                dcc.Graph(
                                    id="admissions-staff-chart"
                                )
                            ]
                        )
                    ]
                )
            ]
        )
    ]
)


# --------------------------------------------------
# 4. UPDATE DASHBOARD WHEN FILTER CHANGES
# --------------------------------------------------

@app.callback(

    Output("kpi-staff", "children"),
    Output("kpi-admissions", "children"),
    Output("kpi-ratio", "children"),
    Output("kpi-pressure", "children"),
    Output("kpi-pressure-value", "children"),

    Output("patients-staff-chart", "figure"),
    Output("staff-distribution-chart", "figure"),
    Output("staff-share-chart", "figure"),
    Output("admissions-staff-chart", "figure"),

    Input("department-filter", "value")
)


def update_dashboard(selected_department):

    # ----------------------------------------------
    # FILTER DATA
    # ----------------------------------------------

    if selected_department == "All Departments":

        filtered = dashboard_df.copy()

    else:

        filtered = dashboard_df[
            dashboard_df["department_name"]
            == selected_department
        ].copy()


    # ----------------------------------------------
    # KPI CALCULATIONS
    # ----------------------------------------------

    total_staff = filtered["Total_Staff"].sum()

    total_admissions = filtered["Admissions"].sum()

    if total_staff > 0:

        patients_per_staff = (
            total_admissions / total_staff
        )

    else:

        patients_per_staff = 0


    highest_row = filtered.loc[
        filtered["Patients_per_Staff"].idxmax()
    ]

    highest_department = highest_row["department_name"]

    highest_value = highest_row["Patients_per_Staff"]


    # ----------------------------------------------
    # CHART 1
    # PATIENTS PER STAFF
    # ----------------------------------------------

    chart1_df = filtered.sort_values(
        "Patients_per_Staff",
        ascending=True
    )

    fig1 = px.bar(

        chart1_df,

        x="Patients_per_Staff",

        y="department_name",

        orientation="h",

        text="Patients_per_Staff",

        labels={
            "Patients_per_Staff":
                "Patients per Staff",

            "department_name":
                "Department"
        },

        title="Patients per Staff by Department"
    )

    fig1.update_traces(

        marker_color="#E67E22",

        texttemplate="%{text:.2f}",

        textposition="outside"
    )

    fig1.update_layout(

        title_x=0.5,

        height=420,

        margin=dict(
            l=20,
            r=30,
            t=60,
            b=40
        ),

        plot_bgcolor="white",

        paper_bgcolor="white"
    )


    # ----------------------------------------------
    # CHART 2
    # TOTAL STAFF
    # ----------------------------------------------

    chart2_df = filtered.sort_values(
        "Total_Staff",
        ascending=False
    )

    fig2 = px.bar(

        chart2_df,

        x="department_name",

        y="Total_Staff",

        text="Total_Staff",

        labels={
            "department_name":
                "Department",

            "Total_Staff":
                "Total Staff"
        },

        title="Staff Distribution Across Departments"
    )

    fig2.update_traces(

        marker_color="#2F5597",

        textposition="outside"
    )

    fig2.update_layout(

        title_x=0.5,

        height=420,

        xaxis_tickangle=-45,

        margin=dict(
            l=40,
            r=30,
            t=60,
            b=100
        ),

        plot_bgcolor="white",

        paper_bgcolor="white"
    )


    # ----------------------------------------------
    # CHART 3
    # STAFF SHARE DONUT
    # ----------------------------------------------

    fig3 = px.pie(

        filtered,

        names="department_name",

        values="Total_Staff",

        hole=0.55,

        title="Staff Share by Department"
    )

    fig3.update_traces(

        textposition="inside",

        textinfo="percent",

        marker=dict(
            line=dict(
                color="white",
                width=1
            )
        )
    )

    fig3.update_layout(

        title_x=0.5,

        height=420,

        margin=dict(
            l=20,
            r=20,
            t=60,
            b=30
        ),

        paper_bgcolor="white"
    )


    # ----------------------------------------------
    # CHART 4
    # ADMISSIONS VS STAFF
    # ----------------------------------------------

    fig4 = px.scatter(

        filtered,

        x="Total_Staff",

        y="Admissions",

        size="Patients_per_Staff",

        hover_name="department_name",

        hover_data={
            "Total_Staff": True,
            "Admissions": True,
            "Patients_per_Staff": ":.2f",
            "Staff_Share_Percent": ":.2f"
        },

        labels={
            "Total_Staff":
                "Total Staff",

            "Admissions":
                "Admissions"
        },

        title="Admissions vs Total Staff"
    )

    fig4.update_traces(

        marker=dict(
            color="#16A085",
            line=dict(width=1)
        )
    )

    fig4.update_layout(

        title_x=0.5,

        height=420,

        margin=dict(
            l=50,
            r=30,
            t=60,
            b=50
        ),

        plot_bgcolor="white",

        paper_bgcolor="white"
    )


    # ----------------------------------------------
    # RETURN EVERYTHING
    # ----------------------------------------------

    return (

        f"{total_staff:,.0f}",

        f"{total_admissions:,.0f}",

        f"{patients_per_staff:.2f}",

        highest_department,

        f"{highest_value:.2f} patients/staff",

        fig1,

        fig2,

        fig3,

        fig4
    )


# --------------------------------------------------
# 5. RUN DASH
# --------------------------------------------------

if __name__ == "__main__":
    app.run(debug=True, port=8050)
```

## Key Visualization Insights

- Patients per Staff is used as the primary staffing workload indicator.
- Departments with higher Patients per Staff indicate relatively higher staffing pressure.
- Staff distribution shows how available staff are allocated across departments.
- Admissions vs Total Staff helps compare patient workload with staffing levels.
- The dashboard provides an interactive Department filter for department-level analysis.

### Note
Staffing pressure is interpreted using the Patients-per-Staff ratio.
This should not be interpreted as confirmed understaffing because the
analysis uses the staffing benchmark data provided for the project.
